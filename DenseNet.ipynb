Image dimensions — e.g. 512×512, 300×400, etc.

Number of channels — RGB, grayscale, RGBA, etc.

Image formats — JPG, PNG, etc.

Pixel-value range

Aspect-ratio distribution

Class-wise image counts — already checked

```python
from PIL import Image
import os
from collections import Counter

dataset_path = "/Data_set/Data set"

classes = ["glioma", "meningioma", "pituitary", "notumor"]

def analyze_images(split):

    sizes = Counter()
    modes = Counter()
    formats = Counter()

    total_images = 0

    for cls in classes:

        folder = os.path.join(dataset_path, split, cls)

        for filename in os.listdir(folder):

            filepath = os.path.join(folder, filename)

            try:
                with Image.open(filepath) as img:

                    total_images += 1

                    # Image dimensions
                    sizes[img.size] += 1

                    # RGB, L (grayscale), RGBA, etc.
                    modes[img.mode] += 1

                    # JPG, PNG, etc.
                    formats[img.format] += 1

            except Exception:
                pass

    return total_images, sizes, modes, formats


for split in ["Training", "Testing"]:

    total, sizes, modes, formats = analyze_images(split)

    print("\n" + "=" * 50)
    print(split)
    print("=" * 50)

    print("Total images:", total)

    print("\nImage modes:")
    for mode, count in modes.items():
        print(f"{mode}: {count}")

    print("\nImage formats:")
    for fmt, count in formats.items():
        print(f"{fmt}: {count}")

    print("\nNumber of unique image sizes:", len(sizes))

    print("\nMost common image sizes:")
    for size, count in sizes.most_common(10):
        print(f"{size}: {count}")
```

Croping the image dataset

```python
!pip install imutils

import cv2
import numpy as np
import imutils
from PIL import Image


def crop_img(img):
    """
    Remove unnecessary background/margins from a brain MRI.

    Based on the preprocessing method provided by the
    dataset author.
    """

    # OpenCV loads images as BGR
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Smooth the image
    gray = cv2.GaussianBlur(gray, (3, 3), 0)

    # Threshold
    thresh = cv2.threshold(
        gray,
        45,
        255,
        cv2.THRESH_BINARY
    )[1]

    # Remove small noise
    thresh = cv2.erode(
        thresh,
        None,
        iterations=2
    )

    thresh = cv2.dilate(
        thresh,
        None,
        iterations=2
    )

    # Find external contours
    cnts = cv2.findContours(
        thresh.copy(),
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    cnts = imutils.grab_contours(cnts)

    # If no contour is found, return original image
    if len(cnts) == 0:
        return img.copy()

    # Largest contour
    c = max(cnts, key=cv2.contourArea)

    # Extreme points
    extLeft = tuple(c[c[:, :, 0].argmin()][0])
    extRight = tuple(c[c[:, :, 0].argmax()][0])
    extTop = tuple(c[c[:, :, 1].argmin()][0])
    extBot = tuple(c[c[:, :, 1].argmax()][0])

    # No additional pixels, same as author's ADD_PIXELS = 0
    ADD_PIXELS = 0

    # Image dimensions
    h, w = img.shape[:2]

    # Make sure crop coordinates stay inside image
    x1 = max(extLeft[0] - ADD_PIXELS, 0)
    x2 = min(extRight[0] + ADD_PIXELS, w)

    y1 = max(extTop[1] - ADD_PIXELS, 0)
    y2 = min(extBot[1] + ADD_PIXELS, h)

    # Crop
    new_img = img[y1:y2, x1:x2].copy()

    return new_img
```

```python
import os
import cv2
from tqdm import tqdm

# Original dataset
dataset_path = "/Data_set/Data set"

# New dataset for cropped images
cropped_path = "/Data_set/Cropped"

classes = [
    "glioma",
    "meningioma",
    "pituitary",
    "notumor"
]

splits = [
    "Training",
    "Testing"
]


for split in splits:

    for cls in classes:

        # Input folder
        input_folder = os.path.join(
            dataset_path,
            split,
            cls
        )

        # Output folder
        output_folder = os.path.join(
            cropped_path,
            split,
            cls
        )

        # Create output folder
        os.makedirs(
            output_folder,
            exist_ok=True
        )

        # Get all files
        files = os.listdir(input_folder)

        print(f"\nProcessing {split}/{cls}")
        print("Images:", len(files))

        # Process every image
        for filename in tqdm(files):

            input_path = os.path.join(
                input_folder,
                filename
            )

            output_path = os.path.join(
                output_folder,
                filename
            )

            # Read image
            image = cv2.imread(input_path)

            # Check if image was readable
            if image is None:
                print("Could not read:", input_path)
                continue

            # Crop margins
            cropped = crop_img(image)

            # Save cropped image
            cv2.imwrite(
                output_path,
                cropped
            )

print("\nCropping completed!")
```

Now we are going to split the dataset

```python
import os
import shutil
from sklearn.model_selection import train_test_split
from tqdm import tqdm

# Cropped dataset
dataset_path = "/Data_set/Cropped"

# Classes
classes = [
    "glioma",
    "meningioma",
    "pituitary",
    "notumor"
]

# Random seed for reproducibility
RANDOM_STATE = 42

# Split ratio
VALIDATION_SIZE = 0.20


for cls in classes:

    # Original cropped training folder
    source_folder = os.path.join(
        dataset_path,
        "Training",
        cls
    )

    # Get image filenames
    files = [
        f for f in os.listdir(source_folder)
        if os.path.isfile(
            os.path.join(source_folder, f)
        )
    ]

    # Split into train and validation
    train_files, val_files = train_test_split(
        files,
        test_size=VALIDATION_SIZE,
        random_state=RANDOM_STATE
    )

    # Create destination folders
    train_folder = os.path.join(
        dataset_path,
        "Train",
        cls
    )

    val_folder = os.path.join(
        dataset_path,
        "Validation",
        cls
    )

    os.makedirs(
        train_folder,
        exist_ok=True
    )

    os.makedirs(
        val_folder,
        exist_ok=True
    )

    # Copy training images
    for filename in tqdm(
        train_files,
        desc=f"Train - {cls}"
    ):

        source = os.path.join(
            source_folder,
            filename
        )

        destination = os.path.join(
            train_folder,
            filename
        )

        shutil.copy2(source, destination)

    # Copy validation images
    for filename in tqdm(
        val_files,
        desc=f"Validation - {cls}"
    ):

        source = os.path.join(
            source_folder,
            filename
        )

        destination = os.path.join(
            val_folder,
            filename
        )

        shutil.copy2(source, destination)


print("\nDataset splitting completed!")
```

```python
import tensorflow as tf
from tensorflow.keras.applications.densenet import preprocess_input

dataset_path = "/Data_set/Cropped"

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
SEED = 42

# --------------------------------
# Load datasets
# --------------------------------

train_ds = tf.keras.utils.image_dataset_from_directory(
    f"{dataset_path}/Train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    f"{dataset_path}/Validation",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    f"{dataset_path}/Testing",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)


# --------------------------------
# Data Augmentation
# Training data ONLY
# --------------------------------

data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomRotation(0.05),
    tf.keras.layers.RandomZoom(0.10),
    tf.keras.layers.RandomTranslation(
        height_factor=0.05,
        width_factor=0.05
    ),
], name="data_augmentation")


# --------------------------------
# Preprocessing / Normalization
# DenseNet121 ImageNet preprocessing
# --------------------------------

def preprocess_train(images, labels):
    images = data_augmentation(images, training=True)
    images = preprocess_input(images)
    return images, labels


def preprocess_val_test(images, labels):
    images = preprocess_input(images)
    return images, labels


# --------------------------------
# Apply preprocessing
# --------------------------------

train_ds = train_ds.map(
    preprocess_train,
    num_parallel_calls=tf.data.AUTOTUNE
)

val_ds = val_ds.map(
    preprocess_val_test,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_ds = test_ds.map(
    preprocess_val_test,
    num_parallel_calls=tf.data.AUTOTUNE
)


# --------------------------------
# Improve performance
# --------------------------------

train_ds = train_ds.prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.prefetch(tf.data.AUTOTUNE)
test_ds = test_ds.prefetch(tf.data.AUTOTUNE)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import DenseNet121


# ============================================================
# 1. Load Pretrained DenseNet121
# ============================================================

base_model = DenseNet121(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)


# ============================================================
# 2. Freeze the DenseNet121 Backbone
# ============================================================

base_model.trainable = False


# ============================================================
# 3. Build the Classification Model
# ============================================================

inputs = layers.Input(shape=(224, 224, 3))

# Feature extraction using pretrained DenseNet121
x = base_model(inputs, training=False)

# Convert feature maps into a single feature vector
x = layers.GlobalAveragePooling2D()(x)

# Reduce overfitting
x = layers.Dropout(0.135545)(x)

# Four output classes
outputs = layers.Dense(
    4,
    activation="softmax"
)(x)

model = Model(inputs, outputs)


# ============================================================
# 4. Display Model Architecture
# ============================================================

model.summary()


# ============================================================
# 5. Compile the Model
# ============================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.00937354
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 6. Callbacks
# ============================================================

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

model_checkpoint = tf.keras.callbacks.ModelCheckpoint(
    "best_densenet121.keras",
    monitor="val_loss",
    save_best_only=True
)


# ============================================================
# 7. Train the Model
# ============================================================

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=[
        early_stopping,
        model_checkpoint
    ]
)
```

Create a dataset function

```python
def create_datasets():

    train_ds = tf.keras.utils.image_dataset_from_directory(
        f"{dataset_path}/Train",
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode="categorical",
        shuffle=True,
        seed=SEED
    )

    val_ds = tf.keras.utils.image_dataset_from_directory(
        f"{dataset_path}/Validation",
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        label_mode="categorical",
        shuffle=False
    )

    # Data augmentation
    data_augmentation = tf.keras.Sequential([
        tf.keras.layers.RandomRotation(0.05),
        tf.keras.layers.RandomZoom(0.10),
        tf.keras.layers.RandomTranslation(
            height_factor=0.05,
            width_factor=0.05
        )
    ])

    # Training preprocessing
    def preprocess_train(images, labels):
        images = data_augmentation(
            images,
            training=True
        )

        images = preprocess_input(images)

        return images, labels

    # Validation preprocessing
    def preprocess_val(images, labels):
        images = preprocess_input(images)

        return images, labels

    # Apply preprocessing
    train_ds = train_ds.map(
        preprocess_train,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    val_ds = val_ds.map(
        preprocess_val,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    # Prefetch
    train_ds = train_ds.prefetch(
        tf.data.AUTOTUNE
    )

    val_ds = val_ds.prefetch(
        tf.data.AUTOTUNE
    )

    return train_ds, val_ds
```

Build DenseNet121

```python
def build_model(config):
    import tensorflow as tf
    from tensorflow.keras import layers, Model
    from tensorflow.keras.applications import DenseNet121

    # Pretrained DenseNet121
    base_model = DenseNet121(
        weights="imagenet",
        include_top=False,
        input_shape=(224, 224, 3)
    )

    # Freeze pretrained layers
    base_model.trainable = False

    # Input
    inputs = layers.Input(
        shape=(224, 224, 3)
    )

    # DenseNet
    x = base_model(
        inputs,
        training=False
    )

    # Global Average Pooling
    x = layers.GlobalAveragePooling2D()(x)

    # ASHA controls this value
    x = layers.Dropout(
        config["dropout"]
    )(x)

    # Four classes
    outputs = layers.Dense(
        NUM_CLASSES,
        activation="softmax"
    )(x)

    model = Model(
        inputs=inputs,
        outputs=outputs
    )

    # ASHA controls learning rate
    optimizer = tf.keras.optimizers.Adam(
        learning_rate=config["learning_rate"]
    )

    model.compile(
        optimizer=optimizer,
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

ASHA training function

```python
def train_model(config):

    print("\n====================================")
    print("Starting new ASHA trial")
    print("Learning rate:", config["learning_rate"])
    print("Dropout:", config["dropout"])
    print("====================================\n")

    # Create datasets INSIDE the trial
    train_ds, val_ds = create_datasets()

    # Build model
    model = build_model(config)

    MAX_EPOCHS = 10

    for epoch in range(MAX_EPOCHS):

        print(f"\nEpoch {epoch + 1}/{MAX_EPOCHS}")

        history = model.fit(
            train_ds,
            validation_data=val_ds,
            epochs=1,
            verbose=1
        )

        # Validation metrics
        val_loss = history.history["val_loss"][0]

        val_accuracy = history.history[
            "val_accuracy"
        ][0]

        # Send results to ASHA
        tune.report({
            "val_loss":val_loss,
            "val_accuracy":val_accuracy
        })
```

Define the search space

```python
search_space = {

    "learning_rate": tune.loguniform(
        1e-4,
        9e-4
    ),

    "dropout": tune.uniform(
        0.1,
        0.5
    )
}
```

Create ASHA

```python
asha_scheduler = ASHAScheduler(
    metric="val_accuracy",
    mode="max",
    max_t=10,
    grace_period=3,
    reduction_factor=2
)
```

Create the tuner

```python
tuner = tune.Tuner(
     tune.with_resources(
        train_model,
        resources={"cpu":2,"gpu": 1}
    ),
    tune_config=tune.TuneConfig(
        scheduler=asha_scheduler,
        num_samples=15
    ),

    param_space=search_space
)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
from tensorflow.keras.applications import DenseNet121


# ============================================================
# 1. Load Pretrained DenseNet121
# ============================================================

base_model = DenseNet121(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)


# ============================================================
# 2. Freeze Backbone Initially
# ============================================================

base_model.trainable = False


# ============================================================
# 3. Build Model
# ============================================================

inputs = layers.Input(shape=(224, 224, 3))

x = base_model(inputs, training=False)

x = layers.GlobalAveragePooling2D()(x)

x = layers.Dropout(0.373234)(x)

outputs = layers.Dense(
    4,
    activation="softmax"
)(x)

model = Model(inputs, outputs)


# ============================================================
# 4. Stage 1: Train Classification Head
# ============================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.000638801
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

checkpoint = tf.keras.callbacks.ModelCheckpoint(
    "best_head.keras",
    monitor="val_loss",
    save_best_only=True
)

history_head = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=10,
    callbacks=[
        early_stopping,
        checkpoint
    ]
)


# ============================================================
# 5. Stage 2: Fine-Tune DenseNet121
# ============================================================

base_model.trainable = True

# Freeze most of DenseNet121
# Only the last 50 layers will be trainable

for layer in base_model.layers[:-80]:
    layer.trainable = False

# Last 50 layers are trainable
for layer in base_model.layers[-80:]:
    layer.trainable = True

# Keep BatchNormalization layers frozen
# This is generally safer when fine-tuning with a
# relatively small medical-image dataset.

for layer in base_model.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False


# ============================================================
# 6. Recompile with MUCH Smaller Learning Rate
# ============================================================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)


# ============================================================
# 8. Fine-Tune
# ============================================================

history_finetune = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=15
)


# ============================================================
# 9. Final Model Summary
# ============================================================

model.summary()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# ============================================================
# 1. Class names
# ============================================================

class_names = [
    "glioma",
    "meningioma",
    "notumor",
    "pituitary"
]

# ============================================================
# 2. Evaluate model on untouched TEST dataset
# ============================================================

test_loss, test_accuracy = model.evaluate(
    test_ds,
    verbose=1
)

print("\n======================================")
print("OVERALL TEST RESULTS")
print("======================================")
print(f"Test Loss:     {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy:.4%}")

# ============================================================
# 3. Get true labels and predictions
# ============================================================

y_true = []
y_pred = []

for images, labels in test_ds:

    predictions = model.predict(images, verbose=0)

    predicted_classes = np.argmax(predictions, axis=1)
    true_classes = np.argmax(labels.numpy(), axis=1)

    y_true.extend(true_classes)
    y_pred.extend(predicted_classes)

y_true = np.array(y_true)
y_pred = np.array(y_pred)

# ============================================================
# 4. Calculate accuracy independently
# ============================================================

overall_accuracy = np.mean(y_true == y_pred)

print("\n======================================")
print("ACCURACY CHECK")
print("======================================")
print(f"Overall Test Accuracy: {overall_accuracy:.4%}")

# ============================================================
# 5. Confusion Matrix
# ============================================================

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=np.arange(len(class_names))
)

print("\n======================================")
print("CONFUSION MATRIX")
print("======================================")
print(cm)

# ============================================================
# 6. Plot Confusion Matrix
# ============================================================

plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.title("DenseNet121 - Test Confusion Matrix")

plt.tight_layout()
plt.show()

# ============================================================
# 7. Classification Report
# ============================================================

print("\n======================================")
print("CLASSIFICATION REPORT")
print("======================================")

print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        digits=4
    )
)

# ============================================================
# 8. Number of correct / incorrect predictions
# ============================================================

correct = np.sum(y_true == y_pred)
incorrect = np.sum(y_true != y_pred)

print("\n======================================")
print("PREDICTION SUMMARY")
print("======================================")
print(f"Total Test Images: {len(y_true)}")
print(f"Correct Predictions: {correct}")
print(f"Incorrect Predictions: {incorrect}")
```
